```python
import os
import subprocess
from google.colab import userdata

github_token = userdata.get("GH_TOKE")
repo_url = "https://github.com/sonasing566-creator/cs107psp-lab.git"

env = os.environ.copy()
env["GIT_USERNAME"] = "x-access-token"
env["GIT_PASSWORD"] = github_token

askpass = "/tmp/git-askpass.sh"

with open(askpass, "w") as f:
    f.write('#!/bin/sh\ncase "$1" in\n')
    f.write('  *Username*) echo "$GIT_USERNAME" ;;\n')
    f.write('  *Password*) echo "$GIT_PASSWORD" ;;\nesac\n')

os.chmod(askpass, 0o700)
env["GIT_ASKPASS"] = askpass
env["GIT_TERMINAL_PROMPT"] = "0"

result = subprocess.run(
    ["git", "clone", repo_url],
    env=env, capture_output=True, text=True
)

if result.returncode == 0:
    print("Repository cloned successfully!")
else:
    print(result.stderr)
```

# Q56: Colab vs Local Conda Environment

## Google Colab

### Pros
1. Runs in a browser with minimal setup.
2. Integrates with Google Drive.
3. Offers cloud computing resources, including GPUs when available.

### Cons
1. Requires an internet connection.
2. Runtime sessions can disconnect or reset.
3. Computing resources may be limited.

## Local Conda Environment

### Pros
1. Can work offline after installation.
2. Allows control over Python versions and packages.
3. Provides access to local files and computer resources.

### Cons
1. Requires installation and configuration.
2. Uses your computer's CPU and memory.
3. GPU support depends on local hardware and drivers.
