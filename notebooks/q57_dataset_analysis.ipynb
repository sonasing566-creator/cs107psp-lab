```python
import csv
import urllib.request

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

with urllib.request.urlopen(url) as response:
    text = response.read().decode("utf-8")

reader = csv.DictReader(text.splitlines())
rows = list(reader)

print("Total rows:", len(rows))
print("Column names:", reader.fieldnames)
print("First 5 rows:")

for row in rows[:5]:
    print(row)
```

```python
# Calculate mean, minimum, and maximum using pure Python

numeric_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

for col in numeric_columns:
    values = [float(row[col]) for row in rows if row[col] != ""]

    mean_value = sum(values) / len(values)

    print(f"\nColumn: {col}")
    print(f"Mean: {mean_value:.2f}")
    print(f"Minimum: {min(values):.2f}")
    print(f"Maximum: {max(values):.2f}")


# Calculate average petal length for each species

groups = {}

for row in rows:
    species = row["species"]
    petal_length = float(row["petal_length"])

    if species not in groups:
        groups[species] = []

    groups[species].append(petal_length)

print("\nAverage Petal Length by Species:")

for species, values in groups.items():
    average = sum(values) / len(values)
    print(f"{species}: {average:.2f}")
```

```python
import pandas as pd
from io import StringIO

# Load the dataset into a Pandas DataFrame
df = pd.read_csv(StringIO(data_text))

# Display the first five rows
print("First 5 rows:")
display(df.head())

# Display dataset information
print("Dataset Information:")
df.info()

# Display statistical summary
print("Statistical Summary:")
display(df.describe())
```

```python
# 1. Average measurements for each flower species
print("Average measurements by species:")
display(df.groupby("species").mean(numeric_only=True))

# 2. Sort flowers by sepal length (largest first)
print("Flowers sorted by sepal length:")
display(df.sort_values("sepal_length", ascending=False).head(10))
```

```python
import matplotlib.pyplot as plt

# 1. Bar chart: Average petal length by species
average_petals = df.groupby("species")["petal_length"].mean()

average_petals.plot(kind="bar")
plt.title("Average Petal Length by Species")
plt.xlabel("Flower Species")
plt.ylabel("Average Petal Length")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# 2. Histogram: Distribution of sepal length
df["sepal_length"].plot(kind="hist", bins=10, edgecolor="black")

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
import os

# 1. Check missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# 2. Remove rows containing missing values
clean_df = df.dropna()

print("\nRows before cleaning:", len(df))
print("Rows after cleaning:", len(clean_df))

# 3. Create the data folder
os.makedirs("data", exist_ok=True)

# 4. Save the cleaned dataset
clean_df.to_csv("data/cleaned.csv", index=False)

print("\nCleaned dataset saved to data/cleaned.csv")
```
